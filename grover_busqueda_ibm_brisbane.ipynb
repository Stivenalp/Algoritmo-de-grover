# <FONT SIZE=5 COLOR="white"> **Aplicación del Algoritmo de Grover en búsquedas no estructuradas: análisis y comparación de la eficiencia de algoritmos de búsqueda con algoritmos cuánticos.** </FONT>

Las búsquedas en bases de datos no estructuradas o listas no ordenadas son un desafío esencial en la teoría de la computación, debido a la ausencia de un orden que permita aprovechar técnicas de optimización clásicas. En este tipo de problemas, la localización de un elemento que cumple una propiedad específica requiere, en el caso clásico, un análisis exhaustivo de los N elementos presentes en la base de datos, lo que implica un costo computacional promedio de O(N). Este límite se mantiene incluso con enfoques avanzados descritos en trabajos fundamentales de autores como Knuth (1998), quien en The Art of Computer Programming analiza a profundidad las técnicas de búsqueda lineal y sus limitaciones en ausencia de estructura (Knuth, 1998). De manera similar, Cormen, Leiserson, Rivest y Stein (2009) subrayan en Introduction to Algorithms que las búsquedas no estructuradas no pueden beneficiarse de optimizaciones derivadas de estructuras jerárquicas, como árboles de búsqueda o tablas hash, destacando que el rendimiento está limitado por la necesidad de inspeccionar cada elemento de manera secuencial. Por otro lado, Sedgewick (2011) y Wirth (1976) abordan estrategias relacionadas con la eficiencia en algoritmos de búsqueda, pero recalcan que los avances en estructuras de datos no eliminan la barrera inherente de O(N) para problemas sin orden. En este contexto, la computación cuántica ofrece una solución innovadora al problema de búsquedas no estructuradas. El Algoritmo de Grover, introducido en 1996, reduce el tiempo de búsqueda a O(√ N), logrando una aceleración cuadrática en comparación con los algoritmos clásicos. Esto se consigue mediante transformaciones iterativas basadas en la superposición cuántica y la interferencia constructiva, lo que amplifica la probabilidad del estado objetivo y atenúa las demás posibilidades (Grover, 1996). Este avance, sin embargo, enfrenta retos en su implementación práctica, como la preservación de la coherencia cuántica, el diseño de oráculos específicos para cada problema y la necesidad de puertas cuánticas de alta fidelidad. Este proyecto se centra en la implementación del Algoritmo Cuántico de Grover, donde se explorará en simulaciones de computación cuántica con apoyo del lenguaje Qiskit, contribuyendo a una comprensión más profunda de cómo estas técnicas pueden resolver problemas fundamentales de búsqueda con mayor eficiencia. El análisis combinado de estas limitaciones clásicas y los avances cuánticos pone de manifiesto el potencial transformador del Algoritmo de Grover para problemas de búsqueda en bases de datos no estructuradas, aportando una perspectiva novedosa que supera los enfoques tradicionales estudiados por autores como Knuth, Cormen et al., Sedgewick y Wirth.

```python
# --- PARÁMETROS PRINCIPALES ---
n = 9                  
N = 2**n
target_number = 51     
shots = 1014           
use_cloud = False      
use_direct_aer = True 
backend_name = "ibm_quantum_simulator" if use_cloud else None

# --- ITERACIONES DE GROVER ---
iterations = 10

#iterations = math.floor(math.pi / 4 * math.sqrt(N))
hardware_iterations = min(iterations,20)
print(f"Buscando {target_number} en {N} elementos. Iteraciones óptimas: {iterations}, usando: {hardware_iterations}")

# --- REGISTROS CUÁNTICOS ---
search_reg = QuantumRegister(n, name='search')
ancilla = QuantumRegister(1, name='ancilla')
c_reg = ClassicalRegister(n, name='result')
qc = QuantumCircuit(search_reg, ancilla, c_reg)

# --- COMPUERTAS NATIVAS DE BRISBANE (mantenidas) ---

# Usa compuertas estándar en lugar de las nativas Brisbane
"""def custom_h(qc, qubit):
    qc.h(qubit)

def custom_cx(qc, control, target):
    qc.cx(control, target)"""

def custom_h(qc, qubit):
    qc.rz(np.pi, qubit)
    qc.sx(qubit)
    qc.rz(np.pi, qubit)
    qc.rz(-np.pi, qubit) 

def custom_cx(qc, control, target):
    qc.rz(-np.pi/2, target)
    qc.ecr(control, target)
    qc.rz(np.pi/2, target) 

# --- INICIALIZACIÓN ---
for q in range(n):
    custom_h(qc, search_reg[q])
qc.x(ancilla[0])        
custom_h(qc, ancilla[0]) 

qc.barrier()

# --- ORÁCULO ---
def oracle_brisbane():
    oracle_qc = QuantumCircuit(search_reg, ancilla, name="oracle")
    target_bits = format(target_number, f'0{n}b')[::-1] 

    # Aplicar X donde el bit es 0 para marcar el estado objetivo
    for i, bit in enumerate(target_bits):
        if bit == '0':
            oracle_qc.x(search_reg[i])

    
    for i in range(n):
        custom_cx(oracle_qc, search_reg[i], ancilla[0])
        
 #   custom_h(oracle_qc, ancilla[0]) 

    # Revertir X para restaurar el estado
    for i, bit in enumerate(target_bits):
        if bit == '0':
            oracle_qc.x(search_reg[i])

    return oracle_qc


# --- DIFUSOR ---
def diffuser_brisbane():
    qc_diff = QuantumCircuit(search_reg, name="diffuser")
    for q in range(n):
        custom_h(qc_diff, search_reg[q])
    qc_diff.x(search_reg)

   # Esta es la parte modificada:
    custom_h(qc_diff, search_reg[-1])        
    for i in range(n - 1):
        custom_cx(qc_diff, search_reg[i], search_reg[-1])
    custom_h(qc_diff, search_reg[-1])         


    qc_diff.x(search_reg)
    for q in range(n):
        custom_h(qc_diff, search_reg[q])
    return qc_diff

# Construir componentes
oracle = oracle_brisbane()
diffuser = diffuser_brisbane()

# --- ITERACIONES DE GROVER ---
for i in range(hardware_iterations):
    qc.barrier()
    qc.compose(oracle, inplace=True)
    # Después del oráculo:
    qc.z(ancilla[0])

    qc.barrier()
    qc.compose(diffuser, inplace=True)
    print(f"Iteración {i+1} completada")

# --- MEDICIÓN ---
qc.measure(search_reg, c_reg)

# --- EJECUCIÓN ---
backend = None
if use_cloud:
    try:
        service = QiskitRuntimeService()
        backend = service.get_backend(backend_name)
        print(f"Backend en nube: {backend.name}")
    except Exception as e:
        print(f"Error en nube ({e}). Usando local...")
        use_cloud = False

if not use_cloud:
    print("Usando simulación local...")
    


if use_direct_aer or not use_cloud:
    simulator = AerSimulator(method='automatic')
    t_qc = transpile(qc, simulator, optimization_level=1)  
    job = simulator.run(t_qc, shots=shots, memory=False)
    result = job.result()
    counts = result.get_counts(t_qc)
    print("Simulación completada (Aer directo).")

# --- VERIFICACIÓN DEL TARGET EN LOS COUNTS ---
  
    target_bitstring = format(target_number, f'0{n}b')
    print("\n📊 Conteos:")
    for bitstring, count in sorted(counts.items(), key=lambda x: x[1], reverse=True):
        print(f"  {bitstring} : {count}")

    if target_bitstring in counts:
        print(f"\nAparece {target_bitstring} con {counts[target_bitstring]} ocurrencias de {shots} tiros.")
    else:
        print(f"\n No aparece {target_bitstring} en los resultados.")

    initial_layout = list(range(n)) + [n//2]  
else:
    # SamplerV2 (solo si actualizaste paquetes)
    sampler = SamplerV2(backend=backend)
    sampler.options.default_shots = shots
    sampler.options.resilience = {"measure_mitigation": True}
    job = sampler.run([t_qc])
    result = job.result()
    res = result[0]
    counts = res.data.meas.get_counts()
    print("Simulación completada (SamplerV2).")

print("Resultados (top 5):", dict(sorted(counts.items(), key=lambda x: x[1], reverse=True)[:5])) 


target_bitstring = format(target_number, f'0{n}b')
prob_target = counts.get(target_bitstring, 0) / shots * 100
print(f"Probabilidad del target ({target_number}): {prob_target:.2f}%")

# Graficar los resultados
plt.figure(figsize=(12, 5))
plt.bar(counts.keys(), counts.values())
plt.xlabel('Bitstrings')
plt.ylabel('Counts')
plt.title('Distribución de Resultados de la Simulación')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
# --- REGISTROS CUÁNTICOS (define una vez) ---
search_reg = QuantumRegister(n, name='search')
ancilla = QuantumRegister(1, name='ancilla')
c_reg = ClassicalRegister(n, name='result')

# --- COMPUERTAS NATIVAS DE BRISBANE ---
def custom_h(qc, qubit):
    qc.rz(np.pi, qubit)
    qc.sx(qubit)
    qc.rz(np.pi, qubit)
    # Nota: Quité qc.rz(-np.pi, qubit) para evitar fases globales que cancelan

def custom_cx(qc, control, target):
    qc.rz(-np.pi/2, target)
    qc.ecr(control, target)
    qc.rz(np.pi/2, target) 

# --- ORÁCULO (usa f para ocultar target, aplica gates una vez) ---
def oracle_brisbane():
    oracle_qc = QuantumCircuit(search_reg, ancilla, name="oracle")
    
    # Usa f para encontrar target (oculto), compute bits una vez (clásico, single)
    target_states = [i for i in range(N) if f(i)]  
    assert len(target_states) == 1, "f debe marcar single target para Grover simple"
    target = target_states[0]  
    target_bits = format(target, f'0{n}b')[::-1] 

    # Aplicar X donde bit es 0 (una vez, para marcar |11...1>)
    for i, bit in enumerate(target_bits):
        if bit == '0':
            oracle_qc.x(search_reg[i])

    # Fase -1: multiCX directo 
    for i in range(n):
        custom_cx(oracle_qc, search_reg[i], ancilla[0])

    # Revertir X (una vez)
    for i, bit in enumerate(target_bits):
        if bit == '0':
            oracle_qc.x(search_reg[i])

    return oracle_qc

# --- DIFUSOR ---
def diffuser_brisbane():
    qc_diff = QuantumCircuit(search_reg, name="diffuser")
    for q in range(n):
        custom_h(qc_diff, search_reg[q])
    qc_diff.x(search_reg)

    # Multi-CZ efectivo
    custom_h(qc_diff, search_reg[-1])
    for i in range(n - 1):
        custom_cx(qc_diff, search_reg[i], search_reg[-1])
    custom_h(qc_diff, search_reg[-1])

    qc_diff.x(search_reg)
    for q in range(n):
        custom_h(qc_diff, search_reg[q])
    return qc_diff
```

```python
# --- INICIALIZACIÓN ---
qc = QuantumCircuit(search_reg, ancilla, c_reg)
for q in range(n):
    custom_h(qc, search_reg[q])
qc.x(ancilla[0])         
custom_h(qc, ancilla[0]) 

qc.barrier()

# Construir componentes
oracle = oracle_brisbane()
diffuser = diffuser_brisbane()

# --- ITERACIONES DE GROVER ---
for i in range(hardware_iterations):
    qc.barrier()
    qc.compose(oracle, inplace=True)
    # Fix para fase residual
    qc.z(ancilla[0])

    qc.barrier()
    qc.compose(diffuser, inplace=True)
    print(f"Iteración {i+1} completada")

# --- MEDICIÓN ---
qc.measure(search_reg, c_reg)
```

```python
# --- EJECUCIÓN ---
backend = None
if use_cloud:
    try:
        service = QiskitRuntimeService()
        backend = service.get_backend(backend_name)
        print(f"Backend en nube: {backend.name}")
    except Exception as e:
        print(f"Error en nube ({e}). Usando local...")
        use_cloud = False

if not use_cloud:
    print("Usando simulación local...")

if use_direct_aer or not use_cloud:
    simulator = AerSimulator(method='automatic')
    t_qc = transpile(qc, simulator, optimization_level=1)
    job = simulator.run(t_qc, shots=shots, memory=False)
    result = job.result()
    counts = result.get_counts(t_qc)
    print("Simulación completada (Aer directo).")

    # --- VERIFICACIÓN DEL TARGET EN LOS COUNTS ---
    target_bitstring = format(target_number, f'0{n}b')  # '000110011' (MSB first, sin reverse)
    if target_bitstring in counts:
        print(f"\n✅ Aparece {target_bitstring} con {counts[target_bitstring]} ocurrencias de {shots} tiros.")
    else:
        print(f"\n❌ No aparece {target_bitstring} en los resultados.")

    prob_target = counts.get(target_bitstring, 0) / shots * 100
    print(f"Probabilidad del target ({target_number}): {prob_target:.2f}%")
else:
    sampler = SamplerV2(backend=backend)
    sampler.options.default_shots = shots
    sampler.options.resilience = {"measure_mitigation": True}
    job = sampler.run([t_qc])
    result = job.result()
    res = result[0]
    counts = res.data.meas.get_counts()
    print("Simulación completada (SamplerV2).")

print("Resultados (top 5):", dict(sorted(counts.items(), key=lambda x: x[1], reverse=True)[:5]))
```
